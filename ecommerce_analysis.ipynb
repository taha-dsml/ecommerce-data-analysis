```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

n = 500

products = ["Laptop", "Phone", "Tablet", "Headphones",
            "Smartwatch", "Keyboard", "Mouse"]

categories = ["Electronics", "Accessories", "Gadgets"]

cities = ["Srinagar", "Delhi", "Mumbai", "Bangalore",
          "Chennai", "Hyderabad"]

payment_methods = ["UPI", "Credit Card", "Debit Card", "Cash"]

df = pd.DataFrame({
    "Order_ID": np.arange(1001, 1001 + n),

    "Date": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="D"
    ),

    "Product": np.random.choice(products, n),

    "Category": np.random.choice(categories, n),

    "Quantity": np.random.randint(1, 6, n),

    "Unit_Price": np.random.choice(
        [500, 1000, 1500, 5000, 10000, 25000, 50000],
        n
    ),

    "City": np.random.choice(cities, n),

    "Payment_Method": np.random.choice(
        payment_methods, n
    ),

    "Customer_Age": np.random.randint(18, 65, n),

    "Rating": np.random.randint(1, 6, n)
})

df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

df.to_csv("ecommerce_sales.csv", index=False)

print(df.head())
print(df.shape)
print(df.info())
```

```python
revenue = df.groupby("Product")["Total_Sales"].sum()
print('The most to least revenue on each',revenue.sort_values(ascending=False))
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()
print('The city with the most sale is',city_sales.idxmax())
```

```python
df["Payment_Method"].value_counts()
print('The most popular payment method is by', df['Payment_Method'].value_counts().idxmax())
```

```python
category_to_quantity = df.groupby("Category")["Quantity"].sum()
print('The category with the most units sold is',category_to_quantity.idxmax())
```

```python
bins = [18, 25, 35, 45, 55, 65]

labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]

df["Age_Group"] = pd.cut(
    df["Customer_Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
age_spending = df.groupby("Age_Group", observed=True)["Total_Sales"].sum()
age_spending
print('The age group to spend the most is',age_spending.idxmax())
```

```python
product_to_rating = df.groupby("Product")["Rating"].mean()
print('The product with the higest rating is',product_to_rating.idxmax())
```

```python
arr_total_sales = df["Total_Sales"].to_numpy()
print('The mean of total sales is',arr_total_sales.mean())
```

```python
print('The standard deviation for total sales is',arr_total_sales.std())
```

```python
top_10 = np.argsort(arr_total_sales)[-10:][::-1]

print('The 10 largest orders are',arr_total_sales[top_10])
```

```python
arr_quantity = df["Quantity"].to_numpy()

correlation = np.corrcoef(arr_quantity, arr_total_sales)

print(correlation)
```

```python
print('The 25th, 50th and 75th percitiles are', np.percentile(arr_total_sales, [25, 50, 75]))
```
